```python
import yaml

yaml_path = '/content/crack_data/crack_dataset.yaml'

with open(yaml_path, 'r') as file:
    data = yaml.safe_load(file)

# Update paths to the extracted Colab directory
data['path'] = '/content/crack_data/yolo_dataset'
data['train'] = 'images/train'
data['val'] = 'images/val'

with open(yaml_path, 'w') as file:
    yaml.dump(data, file)

print("Updated YAML paths for Colab!")
```

```python
from ultralytics import YOLO
import os

# Define the save directory in Google Drive
drive_results_path = '/content/drive/MyDrive/YOLO_Crack_Training'
os.makedirs(drive_results_path, exist_ok=True)

# Load the nano segmentation model
model = YOLO('yolo11s-seg.pt')

# Train it with save_period=1 to save every epoch
results = model.train(
    data='/content/crack_data/crack_dataset.yaml',
    epochs=90,
    imgsz=640,
    batch=16,
    device=0,
    project=drive_results_path,
    name='results',
    save_period=1,             # Save a checkpoint every 1 epoch
    exist_ok=True              # Overwrite existing folder to keep things clean
)
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Correct path: YOLO saves weights in a 'weights' subfolder
model_path = '/content/drive/MyDrive/YOLO_Crack_Training/results/weights/best.pt'

try:
    model = YOLO(model_path)

    # Specify the path to your test image
    image_path = 'https://thumbs.dreamstime.com/b/broken-road-bridge-big-crack-broken-road-bridge-big-crack-bottom-115263227.jpg'

    # Run inference
    results = model.predict(source=image_path, save=True, conf=0.25)

    # Display the results
    for r in results:
        im_array = r.plot()
        im_rgb = cv2.cvtColor(im_array, cv2.COLOR_BGR2RGB)
        plt.imshow(im_rgb)
        plt.title('Prediction Results')
        plt.axis('off')
        plt.show()
except Exception as e:
    print(f"Error: {e}. Check if training finished and weights exist at {model_path}")
```
